# Problem A: Positional and Label-Based Slicing

```python
import pandas as pd

cars = pd.read_csv("cars.csv")
```

```python
print("Shape of cars:")
print(cars.shape)

print("\nColumn names:")
print(cars.columns.tolist())
```

```python
cars_6_to_10 = cars.iloc[5:10]

cars_6_to_10
```

```python
cars_6_to_10[["Model", "mpg", "cyl", "hp", "gear"]]
```

# Problem B: Model Lookup

```python
toyota = cars[cars["Model"] == "Toyota Corolla"]

toyota
```

```python
pontiac = cars[cars["Model"] == "Pontiac Firebird"]

pontiac = pontiac[["Model", "mpg", "hp", "wt"]]

pontiac
```

# Problem C: Multi-Model Subsetting

```python
selected_cars = cars[
    cars["Model"].isin(["Datsun 710", "Lotus Europa", "Ferrari Dino"])
]

selected_cars = selected_cars[["Model", "mpg", "cyl", "hp", "gear"]]
```

```python
print("Selected cars:")
display(selected_cars)

print("Shape:")
print(selected_cars.shape)
```
